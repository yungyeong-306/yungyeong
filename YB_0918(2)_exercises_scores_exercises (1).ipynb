````python
raw_data = {'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
            'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
            'female': [0, 1, 1, 0, 1],
            'age': [42, 52, 36, 24, 73],
            'preTestScore': [4, 24, 31, 2, 3],
            'postTestScore': [25, 94, 57, 62, 70]}

df = pd.DataFrame(raw_data, columns = ['first_name', 'last_name', 'age', 'female', 'preTestScore', 'postTestScore'])
df
````

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

````python
plt.scatter(x=df.preTestScore, y=df.postTestScore, s=df.age)
plt.title('PreTestScore and PostTestScore')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
````

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

````python
plt.scatter(x=df.preTestScore, y=df.postTestScore, s= df.postTestScore * 4.5, c = df.female)
plt.title('preTestScore and postTestScore')
plt.xlabel('preTestScore')
plt.ylabel('preTestScore')
````

### BONUS: Create your own question and answer it.

What is the score change(postTestScore-preTestScore) for each person?

````python
df['score_gap'] = df.postTestScore - df.preTestScore
plt.bar(df.first_name, df.score_gap)
plt.title('Score Change')
plt.xlabel('Name')
plt.ylabel('Score gap')
plt.show()
````
