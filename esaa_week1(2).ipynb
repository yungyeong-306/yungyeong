###**DataFrame**
**파일 로딩 - pd.read_csv()**

*   pd.read_csv() : 파일 로딩해 DataFrame 으로 반환. csv 는 콤마로 필드 구분,  맨 처음 로우는 칼럼명으로 인지하는 게 디폴트
*   DataFrame.head() : DataFrame 맨 앞 N개의 행 반환. 5개가 디폴트

**DataFrame 조회 - info(), describe()**

*   shape : 행 & 열 크기
*   info() : 데이터 크기 / 타입 / Null 수
*   describe() : 개략적인 데이터 분포 확인 (count는 null이 아닌 데이터 수)
*   DataFrame['칼럼명'] : 특정 칼럼 Series 반환
*   value_counts() : 지정 칼럼 데이터값 건수 반환. dropna 인자(null 포함 여부)
*   series: 인덱스 & 칼럼 하나로 구성

**DataFrame과 리스트, 딕셔너리, 넘파이 ndarray 상호 변환**

1) 넘파이 ndarray, 리스트, 딕셔너리 => DataFrame

*   넘파이 ndarray & 리스트: 칼럼명 X / DataFrame: 칼럼명 O
*   생성 인자 Data: 리스트 or 딕셔너리 or 넘파이 ndarray /
    생성 인자 columns: 칼럼명 리스트 입력
*   (딕셔너리의 경우) 키는 칼럼명, 값은 칼럼 데이터로 변환

2) DataFrame => 넘파이 ndarray, 리스트, 딕셔너리

*   values: DataFrame => 넘파이 ndarray
*   values.tolist() : DataFrame => 리스트
*   to_dict() : DataFrame => 딕셔너리. 인자로 'list' 입력 시 딕셔너리 값 리스트형으로 반환

**DataFrame 칼럼 데이터 세트 생성 & 수정**

*   DataFrame [] 내에 칼럼명 입력 후 생성 & 수정

**DataFrame 데이터 삭제 - drop()**

*   DataFrame.drop(labels=None, axis=0, index=None, columns=None, level=None, inplace=False, errors='raise')

*   labels: 칼럼명(리스트 형태 작성 시 여러 컬럼 삭제 가능) or 인덱스
*   axis: 1은 칼럼 / 0은 로우 삭제
*   이상치 데이터 삭제하는 경우 주로 axis=0 (로우 방향 드롭)
*   inplace: True는 원래 DataFrame 자체 수정 (False면 원본 유지, 새로운 DataFrame 반환)

    ⚠ inplace=True 설정한 채 반환 값을 원래 DataFrame 객체로 할당 X (None 나옴)

###**Index 객체**

*   DataFrame, Series의 레코드를 고유하게 식별하는 객체
*   단일 값 반환, 슬라이싱 가능 (단, 객체 값 변경 불가 ex. indexes[0]=5)

**새로운 index 할당 - reset_index()**

*   새로운 연속 숫자 형 인덱스 할당 & 기존 인덱스는 'index' 칼럼명으로 추가
*   Series에 적용 시 DataFrame 반환 (기존 인덱스가 칼럼으로 추가 -> 칼럼 2개)
*   drop=true 설정 시 기존 인덱스 삭제

###**데이터 셀렉션 & 필터링**

*   ⚠ DataFrame 뒤 []: 칼럼명 or 불린 인덱싱 (특정 칼럼 추출 가능)
*   행 & 열 함께 사용하여 데이터 추출 시 iloc[] or loc[]
*   복합 조건 사용 시 개별 조건 ()로 묶고 and & / or | / Not ~ 사용

**1.  위치 기반 인덱싱 - DataFrame iloc[]**

*   정숫값 or 정수형 슬라이싱 / 팬시 리스트 (불린 X)
*   열 위치에 -1 : 맨 마지막 칼럼

**2.   label 기반 인덱싱 - DataFrame loc[]**

*   loc[인덱스값, 칼럼명]
*   슬라이싱 적용 시 종료 값까지 포함 (일반적으로 종료 값 -1 이지만)

### **정렬, Aggregation 함수, GroupBy**

**1.   DataFrame, Series 정렬 - sort_values()**

*   by: 지정 칼럼으로 정렬 (여러 칼럼으로 정렬 시 리스트 형식 사용)
*   ascending=True: 오름차순 (디폴트)
*   inplace=False: 원본 유지, 새로 정렬한 DataFrame 반환 (디폴트)

**2.   Aggregation - min(), sum(), count(), ...**

*   DataFrame에서 바로 호출하면 모든 칼럼에 적용
*   count(): Null 값 반영 X

**3. groupby()**

*   by: 지정 칼럼으로 groupby
*   agg(): groupby 후 반환된 객체에 서로 다른 aggregation 함수 적용
*   여러 개의 칼럼이 서로 다른 aggregation 함수를 groupby에서 호출할 때는 딕셔너리 형태 사용

###**결손 데이터 처리**

*   결손 데이터는 함수 연산 시 제외

**1.   결손 데이터 확인 - isna()**

*   isna().sum(): 결손 데이터 개수 확인

**2.   결손 데이터 대체 - fillna()**

*   반환 변숫값 지정 or inplace=True 설정 필요

###**apply lambda 식 데이터 가공**

**1.   lambda 식 - lambda x : x // 2 (입력 인자 기반 계산식)**

*   함수 선언 & 함수 내 처리를 한 줄로 변환한 식
*   map(): 입력 인자로 여러 값 사용하는 경우
*   if else 지원. 반환 값 입력 후 if 식 (else는 식 먼저) *ex) 'Child' if ...*
*   else if 지원 X -> else 절 ()에 다시 if else 적용해 사용

**2.   판다스 DataFrame의 lambda 식**

*   apply에 lambda 식 적용 *ex) titanic_df['Age'].apply(lambda x : len(x))*

```python
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : 'Child' if x <=15 else ('Adult' if x <=60
                                                                                    else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```

```python
# 나이에 따라 세분화된 분류를 수행하는 함수 생성.
def get_category(age):
  cat = ''
  if age <= 5: cat = 'Baby'
  elif age <= 12: cat = 'Child'
  elif age <= 18: cat = 'Teenager'
  elif age <= 25: cat = 'Student'
  elif age <= 35: cat = 'Young Adult'
  elif age <= 60: cat = 'Adult'
  else : cat = 'Elderly'

  return cat

# lambda 식에 위에서 생성한 get_category() 함수를 반환값으로 지정.
# get_category(X)는 입력값으로 'Age' 칼럼 값을 받아서 해당하는 cat 반환
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x))
titanic_df[['Age', 'Age_cat']].head()
```
